```python
df_tratamento["perdas_normalizadas"] = df_tratamento["perdas_normalizadas"].replace("?", np.nan)
df_tratamento
```

```python
df_tratamento["perdas_normalizadas"] = pd.to_numeric(df_tratamento["perdas_normalizadas"])
df_tratamento
```

```python
df_tratamento["perdas_normalizadas"].median()
```

```python
df_tratamento["perdas_normalizadas"] = df_tratamento["perdas_normalizadas"].fillna(df_tratamento["perdas_normalizadas"].median())
df_tratamento
```

```python
df_tratamento["preco"] = df_tratamento["preco"].replace("?", np.nan)
df_tratamento["preco"] = pd.to_numeric(df_tratamento["preco"])
```

```python
df_tratamento["potencia"] = df_tratamento["potencia"].replace("?", np.nan)
df_tratamento["potencia"] = pd.to_numeric(df_tratamento["potencia"])
df_tratamento["potencia"] = df_tratamento["potencia"].fillna(df_tratamento["potencia"].median())
```

```python
df_tratamento["numero_portas"] = df_tratamento["numero_portas"].replace("?", np.nan)
df_tratamento["numero_portas"] = df_tratamento["numero_portas"].fillna("quatro")
```

```python
df_tratamento["diametro_cilindro"] = df_tratamento["diametro_cilindro"].replace("?", np.nan)
df_tratamento["diametro_cilindro"] = pd.to_numeric(df_tratamento["diametro_cilindro"])
df_tratamento["diametro_cilindro"] = df_tratamento["diametro_cilindro"].fillna(df_tratamento["diametro_cilindro"].median())
```

```python
df_tratamento["curso_pistao"] = df_tratamento["curso_pistao"].replace("?", np.nan)
df_tratamento["curso_pistao"] = pd.to_numeric(df_tratamento["curso_pistao"])
df_tratamento["curso_pistao"] = df_tratamento["curso_pistao"].fillna(df_tratamento["diametro_cilindro"].median())
```

```python
df_tratamento["rpm_maximo"] = df_tratamento["rpm_maximo"].replace("?", np.nan)
df_tratamento["rpm_maximo"] = pd.to_numeric(df_tratamento["rpm_maximo"])
df_tratamento["rpm_maximo"] = df_tratamento["rpm_maximo"].fillna(df_tratamento["rpm_maximo"].median())
```

```python
df_tratamento.to_csv("concessionaria_carros.csv")
```
