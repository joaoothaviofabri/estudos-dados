```python
import pandas as pd
import numpy as np
df = pd.read_excel('covid.xlsx')
df
```

```python
df_tratamento = df
df_tratamento
```

```python
df_tratamento.info()
```

#### Transformando os valores "-" da coluna "Incidência covid-19 (100 mil hab)" em "NaN" 

```python
df_tratamento['Incidência covid-19 (100 mil hab)'] = df_tratamento['Incidência covid-19 (100 mil hab)'].replace('-', np.nan)
df_tratamento
```

```python
df_tratamento['Incidência covid-19 (100 mil hab)'] = pd.to_numeric(df_tratamento['Incidência covid-19 (100 mil hab)'])
df_tratamento.info()
```

#### Arrumando os valores da coluna "Incidência covid-19 (100 mil hab)"

```python
df_tratamento['Taxa mortalidade (100 mil hab)'] = df_tratamento['Taxa mortalidade (100 mil hab)'].replace('O cálculo requer população >= a 100.000 hab', np.nan)
df_tratamento
```

```python
df_tratamento['Taxa mortalidade (100 mil hab)'] = pd.to_numeric(df_tratamento['Taxa mortalidade (100 mil hab)'])
df_tratamento.info()
```

#### Criando Novas Colunas

```python
df_tratamento['Letalidade (%)'] = np.where(df_tratamento['Casos Acumulados'] == 0, np.nan, ((df_tratamento['Óbitos Acumulados'] / df_tratamento['Casos Acumulados'] * 100)))
df_tratamento
```

```python
df_tratamento[' Casos por 100 mil (recalc)'] = (df_tratamento['Casos Acumulados'] / df_tratamento['População']) * 100000
df_tratamento
```

```python
df_tratamento['Óbitos por 100 mil (recalc)'] = (df_tratamento['Óbitos Acumulados'] / df_tratamento['População']) * 100000
df_tratamento
```

```python
df_tratamento.to_csv('parte_b.csv')
```
